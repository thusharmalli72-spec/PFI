```python
df["current_ratio"]=df["Current_Assets"]/df["Current_Liabilities"]
```

```python
df["current_ratio"]
```

```python
df["ROE"]=(df["Net_Income"]/df["Equity"]*100)
```

```python
df["ROE"]
```

```python
df["EPS"]=df["Net_Income"]/df["Shares_Outstanding"]
```

```python
df["EPS"]
```

```python
min=df["current_ratio"].min()
print("minimum RAtio=",min)
```

```python
min=df["ROE"].min()
print("ROE=",min)
```

```python
max=df["EPS"].max()
print("EPS=",min)
```

```python
plt.figure(figsize=(12,7))
plt.plot(df["Year"],df["EPS"],marker="o")
plt.title("EPS trend")
plt.ylabel("Year")
plt.xlabel("EPS")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,7))
plt.plot(df["Year"],df["ROE"],marker="o")
plt.title("ROE trend")
plt.ylabel("Year")
plt.xlabel("ROE")
plt.grid(True)
plt.show()
```
