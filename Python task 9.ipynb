```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
#Python Task 9: Create Bar Charts, Line Charts, Heatmaps, Budget vs Actual charts,
#and correlation plots using Matplotlib and Seaborn to visualize business performance effectively.
```

```python
df=pd.read_csv("stock_data.csv")
```

```python
df=df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["Open"].mean()
```

```python
df["Open"] = df["Open"].fillna(df["Open"].mean())
```

```python
df["High"] = df["High"].fillna(df["High"].mean())
```

```python
df["Close"] = df["Close"].fillna(df["Close"].mean())
```

```python
df["Volume"] = df["Volume"].fillna(df["Volume"].mean())
```

```python
df.isnull().sum()
```

```python
df["Date"]=pd.to_datetime(df["Date"])
```

```python
df.info()
```

```python
#bar chart
plt.figure(figsize=(19,7))
sns.barplot(x="Date",y="Volume",data=df,width=0.6,color="Black")
plt.title("trading volumn over time")
plt.xlabel("Date")
plt.ylabel("Volumn")
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python
#lineplot
plt.figure(figsize=(19,7))
sns.lineplot(x="Date",y="Close",data=df,color="Black")
plt.title("stock closing")
plt.xlabel("Date")
plt.ylabel("Close")
plt.grid(True)
plt.xticks(rotation=45)
plt.show()
```

```python
correlation=df[["Open", "High", "Low", "Close", "Volume"]].corr()
```

```python
correlation
```

```python
#corr
sns.heatmap(correlation,  
           annot=True,    
           cmap="Reds",  
           cbar=False,   
           square=True,  
           linewidth=2)   
plt.show()
```

```python
df["Budget"] = df["Close"]
df["Actual"] = df["Open"]
```

```python
data={
    "department":["HR", "Finance", "Sales", "Marketing", "IT", "Production", "JSW", "bmw"], 
    "Budget" : [500000,10000, 90000, 60000, 10000,85000,45000, 702000],
    "Actual":[48000, 80000,8100, 65000,23000, 617000, 90000, 200000]}
df = pd.DataFrame(data)
print(df)

# using melt to cover to long data
df_long = df.melt(
    id_vars="department",
    value_vars=["Budget", "Actual"],
    var_name="Type",
    value_name="Amount",
)

# Plot
plt.figure(figsize=(10, 6))
sns.barplot(data=df_long, x="department", y="Amount", hue="Type")
plt.title("Budget vs Actual")
plt.xlabel("Department")
plt.ylabel("Amount")
plt.xticks(rotation=45)
plt.grid(True)
plt.show()
```
