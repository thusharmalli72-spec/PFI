```python
df["Date"]=pd.to_datetime(df["Date"])
```

```python
df=df.sort_values("Date")
df
```

```python
#Calculating moving averages
df["MA_20"]=df["Close"].rolling(20).mean()
df["MA_50"]=df["Close"].rolling(50).mean()
df.head(2)
```

```python
plt.figure(figsize=(17,11))
plt.plot(df["Date"] , df["Close"],label="Closing price",marker="o",color="Black")
plt.plot(df["Date"] ,df["MA_20"],label="Moving avrage of 20 days",marker="o",color="purple")
plt.plot(df["Date"] , df["MA_50"],label="moving average of 50 days",marker="o",color="red")
plt.title("stock price trends with moving avarge")
plt.grid(True)
plt.ylabel("Price")
plt.xlabel("DAte")
plt.legend()
plt.xticks(rotation=35)
plt.show()
```
