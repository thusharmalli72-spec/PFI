```python
Sales_report=df.groupby("Category").agg(
    total_revenue=("Sales","sum"),
    total_profit=("Profit","sum"),
)
Sales_report
```

```python
customer_summaries=df.groupby(["Customer_ID","Customer_Name"]).agg(
    Total_sales=("Sales","sum"),
    net_profit=("Profit","sum"),
    total_transaction=("Transaction_ID","count"),
    avg_iteams=("Quantity","mean"),
)
customer_summaries
```

```python
pd.pivot_table(
    df,
    values="Profit",
    index="City",
    columns="Category",
    aggfunc="sum",
    fill_value=0.0,
)
```
