```python
text_columns = ["Brand", "Model", "Color", "Condition", "Country"]

for col in text_columns:
    df[col] = df[col].astype(str).str.strip()

print("Text columns cleaned successfully!")
```

```python
#Check Invalid Price and Mileage
print("Negative Price values:", (df["Price"] < 0).sum())
print("Negative Mileage values:", (df["Mileage"] < 0).sum())
```

```python
#Basic Financial/Business Analysis
total_sales=df["Price"].sum()
print("Total Sales Value:", total_sales)
```

```python
average_price=df["Price"].mean()
print("Average Price",average_price)
```

```python
highest_price = df["Price"].max()
print("Highest Price:", highest_price)
```

```python
lowest_price = df["Price"].min()
print("Lowest Price:", lowest_price)
```

```python
#Total Number of Cars
total_cars = len(df)
print("Total Number of Cars:", total_cars)
```

```python
#Calculate Average Mileage
average_mileage = df["Mileage"].mean()
print("Average Mileage:", average_mileage)
```

```python
#GROUP BY Analysis
brand_summary = df.groupby("Brand")["Price"].agg(
    Number_of_Cars="count",
    Total_Sales_Value="sum",
    Average_Price="mean"
).sort_values("Total_Sales_Value", ascending=False)

brand_summary
```

```python
#Sales Summary by Country
country_summary = df.groupby("Country")["Price"].agg(
    Number_of_Cars="count",
    Total_Sales_Value="sum",
    Average_Price="mean"
).sort_values("Total_Sales_Value", ascending=False)

country_summary
```

```python
#Sales Summary by Condition
condition_summary = df.groupby("Condition")["Price"].agg(
    Number_of_Cars="count",
    Total_Sales_Value="sum",
    Average_Price="mean"
).sort_values("Total_Sales_Value", ascending=False)

condition_summary
```

```python
import matplotlib.pyplot as plt
import seaborn as sns 
```

```python
#Total Sales Value by Brand
plt.figure(figsize=(12, 6))
sns.barplot(x="Brand",y="Price",data=df,width=0.6,color="black")

plt.title("Total Sales Value by Brand")
plt.xlabel("Brand")
plt.ylabel("Total Sales Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 6))

country_summary["Total_Sales_Value"].plot(kind="bar",color="purple")

plt.title("Total Sales Value by Country")
plt.xlabel("Country")
plt.ylabel("Total Sales Value")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))

condition_summary["Average_Price"].plot(kind="bar",color="red")

plt.title("Average Price by Car Condition")
plt.xlabel("Condition")
plt.ylabel("Average Price")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
year_summary = df.groupby("Year")["Price"].sum()

plt.figure(figsize=(12, 6))
sns.lineplot(x="Year",y="Price",data=df,marker="o",color="green")

plt.title("Total Sales Value by Year")
plt.xlabel("Year")
plt.ylabel("Total Sales Value")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df["Price"], bins=40, color="skyblue", edgecolor="black")
plt.title("Price Distribution of Cars")
plt.xlabel("Price")
plt.ylabel("Number of Cars")
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
top_models= df.groupby("Model")["Price"].sum().sort_values(ascending=False)
top_10_models = top_models
top_models
```

```python
plt.figure(figsize=(12, 6))

top_10_models.plot(kind="bar")

plt.title("Top 10 Car Models by Sales Value")
plt.xlabel("Car Model")
plt.ylabel("Total Sales Value")
plt.xticks(rotation=90)
plt.tight_layout()
plt.show()
```

```python
condition_sales = df.groupby("Condition")["Price"].sum()
plt.figure(figsize=(10, 5))
condition_sales.sort_values().plot(kind="barh", color="black")
plt.title("Sales Value by Car Condition")
plt.xlabel("Total Sales Value")
plt.ylabel("Condition")
plt.tight_layout()
plt.show()
```

```python
condition_sales = df.groupby("Condition")["Price"].sum()

plt.figure(figsize=(8, 8))

plt.pie(
    condition_sales,
    labels=condition_sales.index,
    autopct="%1.1f%%"
)

plt.title("Sales Distribution by Car Condition")
plt.tight_layout()
plt.show()
```

```python
print("========== FINANCIAL DASHBOARD KPIs ==========")

print("Total Sales Value :", total_sales)
print("Average Price     :", average_price)
print("Highest Price     :", highest_price)
print("Lowest Price      :", lowest_price)
print("Total Cars        :", total_cars)
print("Average Mileage   :", average_mileage)
```

```python
top_brand = brand_summary["Total_Sales_Value"].idxmax()
top_brand_sales = brand_summary["Total_Sales_Value"].max()

print("Top Brand:", top_brand)
print("Sales Value:", top_brand_sales)
```

```python
top_country = country_summary["Total_Sales_Value"].idxmax()
top_country_sales = country_summary["Total_Sales_Value"].max()

print("Top Country:", top_country)
print("Sales Value:", top_country_sales)
```

```python
top_year = year_summary.idxmax()
top_year_sales = year_summary.max()

print("Year with Highest Sales Value:", top_year)
print("Sales Value:", top_year_sales)
```
