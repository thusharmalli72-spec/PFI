```python
#Task 3 – Financial Calculator Module
#Create reusable Python functions to calculate Simple Interest and Compound Interest. 
#Use loops to automate repetitive financial calculations
```

```python
def simple_intrest(p,r,t):
    si=(p*r*t)/100
    return si
```

```python
def compund_intrest(p,r,t):
    amount=p*(1+r/100)**t
    ci=amount-p
    return ci
```

```python
n=int(input("enter the number of customers:"))
for i in range(n):
    p=int(input("enter the principle amount:"))
    r=int(input("enter the rate:"))
    t=int(input("enter the time"))

    si=simple_intrest(p,r,t)
    ci=compund_intrest(p,r,t)
    print("simple_intrest:",si)
    print("compund_intrest:",ci)
```

```python
#Task 4 – Transaction Automation Module
#Process multiple financial transactions using loops and implement exception handling using try-except blocks to
#validate financial inputs and generate Monthly Bank Reconciliation reports.
```

```python
try:
    balance=int(input("enter the amount"))
    no_of_transaction=int(input("enter the number of transaction"))
    transaction=[]
    for i in range(no_of_transaction):
        try:
            print("no of customers",i+1)
            amount=int(input("enter the amount"))
            transaction_type=input("debit/credit")
            transaction.append([amount,transaction_type])
        except ValueError:
            print("invaild input")
            
    for j in transaction:
        amount=j[0]
        transaction_type=j[1]
        if transaction_type.lower()=="debit":
            balance=balance-amount
        elif transaction_type.lower()=="credit":
            balance=amount+balance
        else:
            print("invaild input")
           
    print("Balance:",balance)
except ValueError:
    print("invaild input")
```

```python
df["Date"]=pd.to_datetime(df["Date"])
```

```python
x=df[["Advertising_Expense","Discount_Percent","Customers"]]
y=df["Sales"]
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train,y_train)
```

```python
y_predict=model.predict(x_test)
```

```python
data=pd.DataFrame({"Acutal sales":y_test,
                   "predict sales":y_predict})
```

```python
MSE=mean_squared_error(y_test,y_predict)
```
